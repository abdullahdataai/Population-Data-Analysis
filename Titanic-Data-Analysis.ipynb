```python
df.columns = df.columns.str.strip()
```

```python
df["Общее население"] = df["Общее население"].astype(str).str.replace(",", "")
df["Общее население"] = pd.to_numeric(df["Общее население"], errors="coerce")
```

```python
df["Общее население"].mean()
df["Общее население"].median()
```

```python
df["городское население"] = df["городское население"].astype(str).str.replace(",", "")
df["городское население"] = pd.to_numeric(df["городское население"], errors="coerce")
```

```python
df["городское население"].mean()
df["городское население"].median()

```

 Q:

 What conclusions can be drawn based on the calculated mean and median values?
By comparing the mean and median values, we can conclude that the population data is not evenly distributed.
In most cases, the mean is higher than the median, which means that a few regions with very large populations (big cities) increase the average value.
This shows that:
•	Population is not equally distributed across regions.
•	A small number of large cities strongly affect the overall statistics.
•	Most regions have lower population compared to big urban centers.

What trends can be identified in the distribution of urban and rural population?
From the analysis of urban and rural population, we can observe clear trends:
•	Urban population is concentrated in major cities.
•	Rural population is spread across many regions but with lower numbers in each area.
•	There is a clear tendency for people to live in cities rather than rural areas.
•	Large cities contain a significant share of the total population
